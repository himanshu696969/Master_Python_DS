```python
daily_steps = [8500, 9200, 9800, 10500, 11000, 8700, 12000]

day = 0

while day < len(daily_steps):
    if daily_steps[day] > 10000:
        print("first day crossing 10,000 steps", day + 1)
        print("steps:", daily_steps[day])
        break

    day += 1
```

```python
cart_prices = [450, 0, 750, 300, 800, 500]

total = 0

for price in cart_prices:
    if price == 0:
        continue
    total += price

    if total > 2000:
        break

print("final total ",total )
```
